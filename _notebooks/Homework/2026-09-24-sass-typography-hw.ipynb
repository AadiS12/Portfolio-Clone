## Popcorn Hack

Click **Run** to see a heading, subheading, and paragraph using semantic tags plus OCS typography classes.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack — Semantic HTML Typography | interactive: true -->
<!-- h1 > h2 > p shows the importance order of the text. -->
<h1 class="ocs_section-title">Semantic HTML</h1>
<h2>builds clear structure</h2>
<p class="ocs_lead">headings first, details after</p>
```

## Homework Hack

Click **Run** to see step-by-step instructions as a card with a heading, paragraphs, and a numbered list.

```python
%%html
<!-- UI_RUNNER: Homework Hack — How to Make a Sandwich | interactive: true -->
<!-- A card groups the whole set of instructions. -->
<!-- An ordered list (ol) fits steps because the order matters. -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Make a Sandwich</h2>
  <h3>Hungry? Here is how to…</h3>
  <p class="ocs_description">make a quick sandwich in just a few easy steps.</p>
  <p class="ocs_text"><strong>Follow these steps:</strong></p>
  <ol>
    <li>place two slices of bread on a plate</li>
    <li>add your favorite filling to one slice</li>
    <li>top it with the other slice and cut it in half</li>
  </ol>
  <h3>Enjoy your sandwich! :)</h3>
</div>
```
