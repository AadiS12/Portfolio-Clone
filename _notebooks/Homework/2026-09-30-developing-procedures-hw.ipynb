## Python 3.13 Developing Procedures

The lesson lists three practice exercises and a procedure-naming mini-game, so this notebook completes all four.

## Practice 1: Rename Procedures Clearly

Each name says exactly what the procedure does, so `make_dessert()` reads like a plan.

````python
# CODE_RUNNER: 3.13 Practice 1 - descriptive procedure names
def measure_flour():
    print("Measuring 2 cups of flour.")

def mix_dessert_ingredients():
    print("Mixing flour, sugar, and eggs.")

def bake_dessert():
    print("Baking at 350°F for 25 minutes.")

def make_dessert():
    measure_flour()
    mix_dessert_ingredients()
    bake_dessert()

make_dessert()
````

## Practice 2: Equivalent Java Methods

```java
public static void makeDessert() {
    measureFlour();
    mixDessertIngredients();
    bakeDessert();
}

private static void measureFlour() {
    System.out.println("Measuring 2 cups of flour.");
}

private static void mixDessertIngredients() {
    System.out.println("Mixing flour, sugar, and eggs.");
}

private static void bakeDessert() {
    System.out.println("Baking at 350°F for 25 minutes.");
}
```

Java uses `camelCase` names and requires `static` here because these methods are called without creating an object. `public` lets other classes call `makeDessert()`, while the helpers are `private` because only this class needs them.

## Practice 3: Split a Long Procedure

Small helpers that return values can each be tested on their own.

````python
# CODE_RUNNER: 3.13 Practice 3 - decompose a long procedure
def choose_ingredient():
    return "flour"

def mix(item):
    return "Mixing " + item + " into the bowl."

def heat():
    return "Heating the oven to 350°F."

def finish():
    return "Dessert is ready!"

def prepare_dessert():
    return [mix(choose_ingredient()), heat(), finish()]

for step in prepare_dessert():
    print(step)

# Test the helpers one at a time.
print("mix test:", mix("sugar") == "Mixing sugar into the bowl.")
print("heat test:", heat() == "Heating the oven to 350°F.")
````

## Mini-Game Practice: Procedure Naming

The original uses `input()`. This version picks an ingredient at random so it runs in the portfolio without waiting for typing.

````python
# CODE_RUNNER: 3.13 Mini-game - why clear procedure names matter
import random

def choose_ingredient():
    return random.choice(["flour", "sugar", "eggs", "milk"])

def prepare_ingredient():
    ingredient = choose_ingredient()
    print("Mixing", ingredient, "into the bowl.")
    print("Heating the oven...")
    print("Dessert is ready!")

# prepare_ingredient() explains itself; a name like do_step() would not.
prepare_ingredient()
````
