## Popcorn Hack 1: Predict, Run, and Fix the Crash

The list has 4 items, so the valid indexes are `0` to `3` (or `-1` to `-4`). Asking for index `4` would crash with an `IndexError`, so the last item is read with `3` or `-1`.

```python
# CODE_RUNNER: Lists Popcorn 1 - predict, run, fix the crash
foods = ["pizza", "sushi", "tacos", "ramen"]

print(foods[0])             # prediction: pizza
print(foods[-1])            # prediction: ramen
print(len(foods))           # prediction: 4
print(foods[len(foods)-1])  # last valid index is len - 1; prediction: ramen
```

## Popcorn Hack 2: Sum and Average

A running total, then the average. An empty list is checked first so we never divide by zero.

```python
# CODE_RUNNER: Lists Popcorn 2 - sum and average
numbers = [7, 6, 8, 5, 7, 9]

total = 0
for n in numbers:
    total += n

print("Total:", total)
print("Average:", total / len(numbers))

sleep = []
if len(sleep) == 0:
    print("No data yet")
else:
    print("Sleep average:", sum(sleep) / len(sleep))
```

## Popcorn Hack 3: Count Then Filter

```python
# CODE_RUNNER: Lists Popcorn 3 - count then filter
seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]

counts = {}
for s in seasons:
    counts[s] = counts.get(s, 0) + 1

popular = [name for name, n in counts.items() if n >= 2]

print(counts)
print(popular)
```

## Homework Hack: Indexing, Accumulator, Count, and Filter

One dataset (movie night votes and ticket prices) used for all four skills.

```python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter on one dataset
votes = ["comedy", "action", "comedy", "horror", "action", "comedy",
         "drama", "action", "comedy", "horror", "comedy", "action"]
prices = [8, 10, 8, 12, 10, 8, 9, 10, 8, 12, 8, 10]

# Indexing
print("Votes cast:", len(votes))
print("Last vote:", votes[-1])

# Accumulator
total = 0
for p in prices:
    total += p
print("Total:", total)
print("Average:", round(total / len(prices), 2))

# Count
counts = {}
for v in votes:
    counts[v] = counts.get(v, 0) + 1
print(counts)

# Filter: genres with at least 3 votes
popular = []
for genre in counts:
    if counts[genre] >= 3:
        popular.append(genre)
print(popular)
```
