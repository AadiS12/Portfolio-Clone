## Python 3.08 Iterations

`for` loops with `continue`, `break`, counters, running totals, and thresholds.

## Popcorn Hack: Wildfire Dispatch Scanner

```python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner
alerts = [1, 4, 2, 6, 3, 9, 5]
safe_cutoff = 3        # below this: skip
critical_cutoff = 9    # at or above this: stop scanning

safe_count = 0
action_count = 0
action_total = 0
critical_found = False

for alert in alerts:
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue                      # skip the rest of this pass

    action_count += 1
    action_total += alert
    print("Alert " + str(alert) + ": action needed")

    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break                         # leave the loop entirely

print("Safe alerts skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(action_total))
print("Critical found: " + str(critical_found))
```

## Homework Hack: Network Latency Monitor

Nine readings, a safe threshold and a critical threshold, `continue`, `break`, and a final report.

```python
# CODE_RUNNER: Homework Hack - themed loop-based checker
latencies = [18, 52, 31, 77, 45, 64, 29, 118, 59]   # milliseconds
safe_cutoff = 40
critical_cutoff = 100

checked = 0
skipped = 0
flagged = 0
flagged_total = 0
critical_found = False

for ms in latencies:
    checked += 1

    if ms < safe_cutoff:
        skipped += 1
        print(str(ms) + " ms: fine, no action")
        continue

    flagged += 1
    flagged_total += ms
    print(str(ms) + " ms: needs attention")

    if ms >= critical_cutoff:
        critical_found = True
        print("Critical latency found. Stop checking.")
        break

print("Final Report")
print("Checked: " + str(checked))
print("Skipped: " + str(skipped))
print("Flagged: " + str(flagged))
print("Total flagged latency: " + str(flagged_total))
print("Critical found: " + str(critical_found))
```
