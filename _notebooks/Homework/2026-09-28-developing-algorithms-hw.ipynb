## Popcorn Hack 1: Translate Pseudocode to Python

The loop adds `3 + 5 + 2 + 4 = 14`, and `14 / 4 = 3.5`.

````python
# CODE_RUNNER: 3.09 Popcorn 1 - pseudocode to Python
hours = [3, 5, 2, 4]

total = 0
for h in hours:
    total += h

average = total / len(hours)

print(total)
print(average)
````

## Popcorn Hack 2: Predict, Then Fix

**Prediction before fixing:** `1`. With `count > capacity`, only `15` counts; the workshop with exactly `12` is full but is missed.  
**Fix:** use `>=`, which counts both full workshops and prints `2`.

````python
# CODE_RUNNER: 3.09 Popcorn 2 - fix the full-workshop counter
signed_up = [12, 7, 15, 9]
capacity = 12

full = 0
for count in signed_up:
    if count >= capacity:      # exactly at capacity is also full
        full += 1

print("Full workshops:", full)
````

## Popcorn Hack 3: Modify a Counter into a Collector

Instead of counting, collect the names of workshops that still have open seats.

````python
# CODE_RUNNER: 3.09 Popcorn 3 - counter into collector
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]
capacity = 12

open_workshops = []
for name, count in zip(workshops, signed_up):
    if count < capacity:
        open_workshops.append(name)

print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
````

## Homework Hack: Supply Closet Restock Report

Three known algorithms combined: collect the shortages, sum them with a loop, and find the largest. Items exactly at target are not short.

````python
# CODE_RUNNER: 3.09 Homework - Safe Passage Heals supply restock report
items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]
target = [30, 20, 25, 15, 30]

# 1) Collect: every item below target, with how many are missing.
short_items = []
short_amounts = []
for i in range(len(items)):
    missing = target[i] - in_stock[i]
    if missing > 0:
        short_items.append(items[i])
        short_amounts.append(missing)

# 2) Sum the shortages with a loop (no sum()).
total_needed = 0
for amount in short_amounts:
    total_needed += amount

# 3) Find the largest shortage and remember where it is.
worst = 0
for i in range(len(short_amounts)):
    if short_amounts[i] > short_amounts[worst]:
        worst = i

# 4) Report, including the fully-stocked case.
if len(short_items) == 0:
    print("Fully stocked!")
else:
    print("Restock needed:")
    for i in range(len(short_items)):
        print(short_items[i] + ": need " + str(short_amounts[i]))
    print("Total items needed:", total_needed)
    print("Most urgent: " + short_items[worst] + " (short by " + str(short_amounts[worst]) + ")")
````
