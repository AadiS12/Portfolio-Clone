## OCS Container Grammar

The finished [OCS Containers lesson](https://pages.opencodingsociety.com/navigation/sass/containers/lesson) hacks are below. Click **Run** on each labeled UI runner to preview the layout. Both solutions rely only on the shared OCS container grammar, with no custom CSS.

## Popcorn Hack: Parade Invite Rescue

The broken invite is rebuilt with a single `ocs__container` wrapping a single `ocs__card`. The details sit in a responsive 3-column standard grid, and the start time is the accented cell.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack — Parade Invite Rescue | interactive: true -->
<!-- One container centers the page content and keeps the width responsive. -->
<!-- One card holds the heading, blurb, and the three-part details grid. -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">You're Invited: Poway Rotary Parade</h2>
    <p class="ocs__description">Bring your family and neighbors out to cheer on floats, bands, and local service clubs.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell">Where: Poway Road, downtown</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Kickoff: 9:00 AM sharp</div>
      <div class="ocs__grid-cell">Perks: Free snacks and live music</div>
    </div>
  </div>
</div>
```

## Homework Hack: Parade Hub Page

The hub page keeps to exactly one container and one card. Inside are the required standard `cols-3` grid (header, accent, and muted cells), an OCS table with two body rows, and a closing callout.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Poway Rotary Parade Hub | interactive: true -->
<!-- Everything lives in one container/card pair so spacing stays uniform. -->
<!-- The grid lists headline facts; the table tracks what is ready for parade day. -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Rotary Parade Volunteer Hub</h2>
    <p class="ocs__description">One page for volunteers, marchers, and spectators to find the plan for parade morning.</p>

    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Parade Morning Plan</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Step-Off: 9:00 AM from Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Crew Arrival: 7:45 AM at the staging lot</div>
    </div>

    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <tr><th>Task</th><th>Current Status</th></tr>
        <tr><td>Float Decorating</td><td>Finished</td></tr>
        <tr><td>Canned Food Collection</td><td>Boxes still needed</td></tr>
      </table>
    </div>

    <div class="ocs__callout">Wrapping the hub in one OCS container keeps it tidy on both phones and desktops.</div>
  </div>
</div>
```
