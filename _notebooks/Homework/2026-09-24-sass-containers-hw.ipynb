## OCS Container Grammar

Both hacks below use the shared OCS container grammar (`ocs__container` > `ocs__card` > grid, table, callout) with no custom CSS. Click **Run** on each UI runner to preview.

## Popcorn Hack: Parade Invite Rescue

The invite is rebuilt inside one `ocs__container` and one `ocs__card`. The details sit in a responsive 3-column standard grid, with the start time accented.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack — Parade Invite Rescue | interactive: true -->
<!-- One container centers everything; one card groups the invite content. -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Robotics Expo Invite</h2>
    <p class="ocs__description">Come see student-built robots compete, demo, and show off what they have been engineering all semester.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell">Where: Main Gym</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Starts At: 6:30 PM</div>
      <div class="ocs__grid-cell">Bring: Friends, family, and curiosity</div>
    </div>
  </div>
</div>
```

## Homework Hack: Hub Page

Exactly one container and one card hold: a standard `cols-3` grid (header, accent, and muted cells), an OCS table with two body rows, and a closing callout.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Robotics Expo Hub | interactive: true -->
<!-- A single container and a single card wrap the whole hub for consistent spacing. -->
<!-- The grid shows key info, the table tracks team status, the callout closes the page. -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Robotics Expo Hub</h2>
    <p class="ocs__description">Schedules, check-in details, and team progress for expo night, all on one responsive page.</p>

    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Expo Night Details</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Doors Open: 6:30 PM in the Main Gym</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Team Setup: 5:30 PM</div>
    </div>

    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <tr><th>Team</th><th>Status</th></tr>
        <tr><td>Line-Follower Bots</td><td>Ready to demo</td></tr>
        <tr><td>Robot Arm Crew</td><td>Final calibration</td></tr>
      </table>
    </div>

    <div class="ocs__callout">Using one OCS container keeps the whole hub aligned and readable on both phones and laptops.</div>
  </div>
</div>
```
