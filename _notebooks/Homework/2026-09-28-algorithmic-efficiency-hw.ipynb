## Python 3.17 Algorithmic Efficiency

Four Popcorn Hacks, the Knowledge Check, and the linear vs binary search homework.

## Popcorn Hack 1: Does One Method Always Win?

On the sorted list `[3, 6, 9, 12, 15, 18, 21]` with target `3`, linear search finds it on the **1st check**. Binary search checks `12`, then `6`, then `3`, which is **3 checks**. Linear wins here, but only because the target happens to be first.

```python
# CODE_RUNNER: 3.17 Popcorn 1 - count checks for target 3
data = [3, 6, 9, 12, 15, 18, 21]
target = 3

# Linear search
linear_checks = 0
for value in data:
    linear_checks += 1
    if value == target:
        break

# Binary search
low, high = 0, len(data) - 1
binary_checks = 0
while low <= high:
    mid = (low + high) // 2
    binary_checks += 1
    if data[mid] == target:
        break
    elif data[mid] < target:
        low = mid + 1
    else:
        high = mid - 1

print("Linear search checks:", linear_checks)
print("Binary search checks:", binary_checks)
```

## Popcorn Hack 2: Compare Growth

When `n` doubles from `1000` to `2000`, linear worst-case work doubles from `1000` to `2000`. Binary search goes from `10` to `11` checks. For many searches on a large sorted list, binary search is the better choice.

```python
# CODE_RUNNER: 3.17 Popcorn 2 - compare search growth
import math

for n in [1000, 2000]:
    print("List size:", n)
    print("Linear search: up to", n, "checks")
    print("Binary search: up to", math.floor(math.log2(n)) + 1, "checks")
    print()
```

## Popcorn Hack 3: Small Input, Big Difference

At `n = 10`: linear `10`, quadratic `100`, exponential `1024`. At `n = 20`: linear `20`, quadratic `400`, exponential `1,048,576`. Exponential growth explodes fastest, so on a large input I would try the linear approach first.

```python
# CODE_RUNNER: 3.17 Popcorn 3 - compare growth patterns
for n in [10, 20]:
    print("Input size:", n)
    print("  Linear:", n)
    print("  Quadratic:", n ** 2)
    print("  Exponential:", 2 ** n)
```

## Homework Hack: Choose an Algorithm

The list is already sorted, so worst-case counts are compared for `n = 100`, `1000`, and `10000`.

```python
# CODE_RUNNER: 3.17 Homework - search counts for three list sizes
import math

for n in [100, 1000, 10000]:
    print("List size:", n)
    print("  Linear search worst case:", n, "checks")
    print("  Binary search worst case:", math.floor(math.log2(n)) + 1, "checks")
```
