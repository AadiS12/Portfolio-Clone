## Homework Hack: Interactive Settings Panel

Three labeled OCS toggles each control a real feature on the page, and a status line reports how many toggles are currently on.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Interactive Toggle Settings | interactive: true -->
<!-- The OCS toggle classes let SASS style every switch, so no custom CSS is needed. -->
<!-- JavaScript reads .checked because it directly reports whether a checkbox is on. -->

<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Study Dashboard Settings</h2>
    <p class="ocs__description">Turn each switch on or off to change what this panel shows.</p>

    <label class="ocs__toggle ocs__toggle--green">
      <input id="alerts-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Deadline Alerts</span>
    </label>

    <label class="ocs__toggle ocs__toggle--green">
      <input id="quiet-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Quiet Mode</span>
    </label>

    <label class="ocs__toggle ocs__toggle--green">
      <input id="quote-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Motivation Quote</span>
    </label>

    <div class="ocs__callout">Switches turned on: <output id="on-count" aria-live="polite">0</output></div>
    <p id="alerts-message" hidden>⏰ Deadline alerts are active.</p>
    <div id="quiet-panel" class="ocs__card" hidden><strong>Quiet Mode:</strong> Sounds and pop-ups are muted so you can focus.</div>
    <p id="quote-message" hidden>🌟 "Small steps every day add up to big progress."</p>
  </div>
</div>

<script>
  const allSwitches = document.querySelectorAll('.ocs__toggle-input');
  const onCount = document.getElementById('on-count');

  // Count the checked switches from scratch each time so the number can never drift.
  function refreshCount() {
    onCount.textContent = Array.from(allSwitches).filter(function(sw) { return sw.checked; }).length;
  }

  // Wire a switch to the element it shows or hides.
  function bindToggle(toggleId, targetId) {
    const target = document.getElementById(targetId);
    document.getElementById(toggleId).addEventListener('change', function(event) {
      target.hidden = !event.target.checked;
      refreshCount();
    });
  }

  bindToggle('alerts-toggle', 'alerts-message');
  bindToggle('quiet-toggle', 'quiet-panel');
  bindToggle('quote-toggle', 'quote-message');
</script>
```
