## Homework Hack: Interactive Settings Panel

Three labeled OCS toggles. Each one shows or hides a real element, and a counter shows how many are on.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Interactive Toggle Settings | interactive: true -->
<!-- The OCS toggle classes provide the switch styling; JavaScript only reads .checked. -->

<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Study Dashboard Settings</h2>
    <p class="ocs__description">Flip a switch to show or hide a feature below.</p>

    <label class="ocs__toggle ocs__toggle--green">
      <input id="banner-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Welcome Banner</span>
    </label>

    <label class="ocs__toggle ocs__toggle--green">
      <input id="progress-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Progress Bar</span>
    </label>

    <label class="ocs__toggle ocs__toggle--green">
      <input id="quote-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Daily Quote</span>
    </label>

    <div class="ocs__callout">Features turned on: <output id="on-count" aria-live="polite">0</output></div>
    <p id="banner" hidden>👋 Welcome back! Ready to study?</p>
    <div id="progress-panel" class="ocs__card" hidden>Weekly goal: <progress value="60" max="100"></progress> 60%</div>
    <p id="quote" hidden>“Small steps every day add up.”</p>
  </div>
</div>

<script>
  const switches = document.querySelectorAll('.ocs__toggle-input');
  const onCount = document.getElementById('on-count');

  // Recount checked switches so the number always matches the screen.
  function refreshCount() {
    onCount.textContent = Array.from(switches).filter(function(s) { return s.checked; }).length;
  }

  // Wire a toggle to the element it should show or hide.
  function link(toggleId, targetId) {
    document.getElementById(toggleId).addEventListener('change', function(event) {
      document.getElementById(targetId).hidden = !event.target.checked;
      refreshCount();
    });
  }

  link('banner-toggle', 'banner');
  link('progress-toggle', 'progress-panel');
  link('quote-toggle', 'quote');
</script>
```
