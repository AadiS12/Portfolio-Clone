## Popcorn Hack 1: Read the API, Then Use It

The jersey range is changed to `1` to `55`, and the square court has area `225`, so its side is `15.0`. The random game and jersey change on every run.

````python
# CODE_RUNNER: 3.14 Popcorn 1 - use math and random APIs
import math
import random

print("random.randint:", random.randint.__doc__)
print("math.sqrt:", math.sqrt.__doc__)
print()

points_per_game = [24, 18, 31, 27, 22]

print("Random game to review:", random.choice(points_per_game), "points")
print("Random jersey number (1-55):", random.randint(1, 55))
print("Square court side:", math.sqrt(225), "feet")

average = sum(points_per_game) / len(points_per_game)
print("Points per game:", average, "-> rounded up:", math.ceil(average))
````

## Popcorn Hack 2: Write a Flask Route

The roster is changed and a new `/coach` route is added. Flask is used when installed; otherwise the same data is printed from a dictionary so the cell still runs.

````python
# CODE_RUNNER: 3.14 Popcorn 2 - Flask routes with a fallback
try:
    from flask import Flask, jsonify

    app = Flask(__name__)

    @app.route("/score")
    def score():
        return jsonify({"home": 3, "away": 2, "set": 2})

    @app.route("/roster")
    def roster():
        return jsonify({"players": ["Maya", "Jordan", "Sam"]})

    @app.route("/coach")
    def coach():
        return jsonify({"coach": "Coach Rivera", "focus": "serve return"})

    client = app.test_client()      # visits routes without a browser
    for path in ["/score", "/roster", "/coach"]:
        print(path, "says:", client.get(path).get_json())

except ModuleNotFoundError:
    routes = {
        "/score": {"home": 3, "away": 2, "set": 2},
        "/roster": {"players": ["Maya", "Jordan", "Sam"]},
        "/coach": {"coach": "Coach Rivera", "focus": "serve return"},
    }
    for path, data in routes.items():
        print(path, "says:", data)
````

## Homework Hack: Soccer Season Stats Report

Four standard-library modules (`json`, `statistics`, `math`, `datetime`). The documented procedure returns total goals, average, best match, and goals by opponent.

````python
# CODE_RUNNER: 3.14 Homework - soccer season stats report
import json
import math
import statistics
from datetime import date

matches = [
    {"date": "2026-09-05", "opponent": "Westview", "goals": 2, "minutes": 90},
    {"date": "2026-09-12", "opponent": "Del Norte", "goals": 4, "minutes": 95},
    {"date": "2026-09-19", "opponent": "Westview", "goals": 1, "minutes": 90},
    {"date": "2026-09-26", "opponent": "Poway", "goals": 3, "minutes": 100},
]

def summarize_season(season):
    """Summarize a list of match dictionaries.

    Parameters:
        season: list of dicts with date, opponent, goals, and minutes.
    Returns:
        dict with total_goals, average_goals, best_match, and goals_by_opponent.
    """
    goals = [m["goals"] for m in season]
    by_opponent = {}
    for m in season:
        by_opponent[m["opponent"]] = by_opponent.get(m["opponent"], 0) + m["goals"]
    return {
        "total_goals": sum(goals),
        "average_goals": round(statistics.mean(goals), 2),
        "best_match": max(goals),
        "goals_by_opponent": by_opponent,
    }

report = summarize_season(matches)

# datetime: weekday of the first match; math: total minutes rounded up to whole hours
report["first_match_weekday"] = date.fromisoformat(matches[0]["date"]).strftime("%A")
report["hours_played"] = math.ceil(sum(m["minutes"] for m in matches) / 60)

print(json.dumps(report, indent=2))

requirements_txt = "pandas\nFlask\n"
print("\nrequirements.txt:\n" + requirements_txt)
````
