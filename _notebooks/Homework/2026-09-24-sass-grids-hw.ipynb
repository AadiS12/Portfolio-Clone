## OCS Grid Grammar

Hacks for the [OCS Grids lesson](https://pages.opencodingsociety.com/sass/grids/). Click **Run** on each UI runner to preview. The grid classes come from the portfolio's SASS, so no custom CSS is needed.

## Popcorn Hack 1: Refactor the Beakers

One `ocs__grid ocs__grid--standard` holds all three beakers, and each beaker is an `ocs__grid-cell`.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 — Three OCS Beakers | interactive: true -->
<!-- One standard grid is the parent; each beaker is a child cell. -->
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell">Beaker A: Distilled Water</div>
  <div class="ocs__grid-cell">Beaker B: Sugar Water</div>
  <div class="ocs__grid-cell">Beaker C: Lemon Juice</div>
</div>
```

## Popcorn Hack 2: Size the Control Beaker

The `fr` unit divides free space in proportion. `1fr 1fr 1fr 2fr` makes four columns, and the last one is twice as wide as each of the others. The lesson allows a `grid-template-columns` style for this hack.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 — Control Beaker Twice as Wide | interactive: true -->
<!-- Four fractions = four columns in one row. -->
<!-- The final 2fr gives the control beaker double the width of a 1fr column. -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Trial 1</div>
  <div class="ocs__grid-cell">Trial 2</div>
  <div class="ocs__grid-cell">Trial 3</div>
  <div class="ocs__grid-cell">Control Beaker</div>
</div>
```

## MCQ Check

1. **A: `cols-4`.** Adding `cols-4` to a standard grid sets four columns.
2. **B: two boxes, the first twice as wide.** `2fr 1fr` splits the row 2 to 1.
3. **C: `ocs__grid-cell--header`.** This modifier makes a cell span the whole row.
4. **D: `<div class="my-row" style="display:flex">`.** It invents a class and uses inline layout, which breaks the lesson's rules.

**MCQ Grids: 4/4 \| answers: A,B,C,D**

## Homework Hack: Plant Growth and Calculator

Click **Run** to see PASS or FIX for each lesson rule. Grid 1 has a header cell as the title and an accented outlier (Plant 3). Grid 2 is a calculator grid with an accented Avg key.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Plant Growth and Calculator Grids with Live Checks | interactive: true, render_mode: javascript -->
<!-- Both grids are written as one HTML string so the same text can be checked and displayed. -->
<!-- Grid 1 uses the standard variant; Grid 2 uses the calculator variant. -->
<script>
const myHTML = [
  '<div class="ocs__grid ocs__grid--standard">',
  '  <div class="ocs__grid-cell ocs__grid-cell--header">Plant Growth After 2 Weeks (cm)</div>',
  '  <div class="ocs__grid-cell">Plant 1: 8.2</div>',
  '  <div class="ocs__grid-cell">Plant 2: 8.6</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Plant 3: 14.9</div>',
  '</div>',
  '<br>',
  '<div class="ocs__grid ocs__grid--calculator">',
  '  <div class="ocs__grid-cell">4</div>',
  '  <div class="ocs__grid-cell">5</div>',
  '  <div class="ocs__grid-cell">6</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>',
  '</div>'
].join('');

// Collect every class name that appears in the markup.
let classList = [];
for (const found of (myHTML.match(/class="[^"]*"/g) || [])) {
  classList = classList.concat(found.slice(7, -1).split(' ').filter(Boolean));
}
const madeUp = classList.filter(name => !name.startsWith('ocs__') && !name.startsWith('cols-'));
const countOf = name => classList.filter(c => c === name).length;
function line(label, ok) {
  return '<li>' + (ok ? 'PASS  ' : 'FIX  ') + label + '</li>';
}
const checks = '<ul>' +
  line('No inline style attributes', !myHTML.includes('style=')) +
  line('No made up classes' + (madeUp.length ? ' (found: ' + madeUp.join(', ') + ')' : ''), madeUp.length === 0) +
  line('Has an ocs__grid root', classList.includes('ocs__grid')) +
  line('Uses the standard variant for Grid 1', classList.includes('ocs__grid--standard')) +
  line('Uses the calculator variant for Grid 2', classList.includes('ocs__grid--calculator')) +
  line('Every box uses ocs__grid-cell', countOf('ocs__grid-cell') >= 8) +
  line('Title is a header cell', classList.includes('ocs__grid-cell--header')) +
  line('Has accent cells (outlier plant and Avg key)', countOf('ocs__grid-cell--accent') >= 2) +
  '</ul>';
outputElement.innerHTML = checks + myHTML;
</script>
```
