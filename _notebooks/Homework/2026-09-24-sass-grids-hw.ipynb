## Popcorn Hack 2: Size the Control Beaker

Four `fr` values create four columns in one row, and the final `2fr` makes the Control Beaker twice as wide as any other column. The lesson explicitly allows a `grid-template-columns` style for this `fr` hack.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 — Control Beaker Twice as Wide | interactive: true -->
<!-- Four fractions in grid-template-columns produce four boxes in a row. -->
<!-- 2fr on the last column makes the control twice the width of each 1fr column. -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Trial 1</div>
  <div class="ocs__grid-cell">Trial 2</div>
  <div class="ocs__grid-cell">Trial 3</div>
  <div class="ocs__grid-cell">Control Beaker</div>
</div>
```

## Homework Hack: Temperature Trial and Calculator

Click **Run** to check the editable code and preview both grids. The runner reports **PASS** or **FIX** for each lesson rule. The temperature title is a header cell spanning the standard grid, Beaker 3 is accented as the outlier, and the calculator's Avg key is accented.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Temperature and Calculator Grids with Live Checks | interactive: true, render_mode: javascript -->
<!-- Both grids are written as one HTML string so the same text can be checked and rendered. -->
<!-- Grid 1 flags the outlier beaker; Grid 2 uses the calculator variant with an Avg key. -->
<script>
const gridRows = [
  '<div class="ocs__grid ocs__grid--standard">',
  '  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>',
  '  <div class="ocs__grid-cell">Beaker 1: 20.8</div>',
  '  <div class="ocs__grid-cell">Beaker 2: 21.3</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 29.7</div>',
  '</div>',
  '<br>',
  '<div class="ocs__grid ocs__grid--calculator">',
  '  <div class="ocs__grid-cell">4</div>',
  '  <div class="ocs__grid-cell">5</div>',
  '  <div class="ocs__grid-cell">6</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>',
  '</div>'
];
const myHTML = gridRows.join('');

// Pull every class name out of the markup so the checks can inspect them.
const usedClasses = [];
for (const attr of (myHTML.match(/class="[^"]*"/g) || [])) {
  for (const name of attr.slice(7, -1).split(' ')) {
    if (name) usedClasses.push(name);
  }
}
const notOcs = usedClasses.filter(n => !n.startsWith('ocs__') && !n.startsWith('cols-'));
const countOf = target => usedClasses.filter(n => n === target).length;

function row(label, ok) {
  return '<li>' + (ok ? 'PASS  ' : 'FIX  ') + label + '</li>';
}
const report = '<ul>' +
  row('No inline style attributes', !myHTML.includes('style=')) +
  row('No made up classes' + (notOcs.length ? ' (found: ' + notOcs.join(', ') + ')' : ''), notOcs.length === 0) +
  row('Has an ocs__grid root', countOf('ocs__grid') > 0) +
  row('Uses the standard variant for Grid 1', countOf('ocs__grid--standard') > 0) +
  row('Uses the calculator variant for Grid 2', countOf('ocs__grid--calculator') > 0) +
  row('Every box uses ocs__grid-cell', countOf('ocs__grid-cell') >= 8) +
  row('Title is a header cell', countOf('ocs__grid-cell--header') > 0) +
  row('Has accent cells (outlier beaker and Avg key)', countOf('ocs__grid-cell--accent') >= 2) +
  '</ul>';
outputElement.innerHTML = report + myHTML;
</script>
```
