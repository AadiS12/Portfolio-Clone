## Popcorn Hack 1: Simple Procedure Call and Parameters

`report_status(location, status)` is defined once and called twice with different arguments.

```python
# CODE_RUNNER: Procedures Popcorn #1
def report_status(location, status):
    print("Status update from " + location + ": " + status)

report_status("Poway", "Shelter open")
report_status("Rancho Bernardo", "Supplies delivered")
```

## Popcorn Hack 2: Decomposition and Helper Procedures

`is_priority()` answers one small question and returns a Boolean. `process_report()` calls it and picks the response.

```python
# CODE_RUNNER: Popcorn Hack 2 - Decomposition & Helper Procedures
def is_priority(incident_type):
    return incident_type == "Medical Emergency"

def process_report(location, incident_type):
    if is_priority(incident_type):
        print("PRIORITY: " + incident_type + " reported in " + location)
    else:
        print("STANDARD REPORT: " + incident_type + " reported in " + location)

process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

## Homework Hack: Emergency Response Dispatcher

The helper `check_priority()` returns a Boolean, and `dispatch_response()` calls it to choose the dispatch type.

```python
# CODE_RUNNER: 3.12 Homework - emergency response dispatcher
def check_priority(incident_type):
    priority_types = ["Medical Emergency", "Fire"]
    return incident_type in priority_types

def dispatch_response(location, incident_type):
    if check_priority(incident_type):
        print("PRIORITY DISPATCH: Send responders to " + location + " for " + incident_type)
    else:
        print("STANDARD DISPATCH: Record " + incident_type + " in " + location)

dispatch_response("Poway", "Medical Emergency")
dispatch_response("Rancho Bernardo", "Power Outage")
dispatch_response("Poway", "Fire")
```

## Explanation

The priority rule lives in one helper, so changing it later means editing one place. Parameters let the same procedures handle any location and incident type.
