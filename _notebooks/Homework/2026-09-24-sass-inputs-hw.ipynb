## Popcorn Hack

`ocs__input` is the base class. `large`, `small`, and `medium` set the size, and `gradient` adds the gradient style. Click **Run** to preview.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack — Input Grammar | interactive: true -->
<!-- The base class plus modifiers replaces any custom CSS. -->
<input type="text" class="ocs__input large gradient" placeholder="Search library books...">
<input type="text" class="ocs__input small" placeholder="Student First Name">
```

## Homework Hack

Three signup fields styled only with OCS classes. The email field uses `type="email"` so browsers validate it.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Club Signup Inputs | interactive: true -->
<!-- Each input is ocs__input plus one size modifier; the first also gets the gradient look. -->
<input type="text" class="ocs__input large gradient" placeholder="Club Interest">
<input type="text" class="ocs__input small" placeholder="Member First Name">
<input type="email" class="ocs__input medium" placeholder="Member Email">
```
