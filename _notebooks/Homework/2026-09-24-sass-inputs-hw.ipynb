## Popcorn Hack

`ocs__input` is the base class. The `large`, `small` and `medium` modifiers set the size, and `gradient` adds the gradient style. No custom CSS is used. Click **Run** to preview.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack — Input Grammar | interactive: true -->
<!-- Base class plus modifiers replaces any custom CSS. -->
<input type="text" class="ocs__input large gradient" placeholder="Search the recipe box...">
<input type="text" class="ocs__input small" placeholder="Chef Nickname">
```

## Homework Hack

Three sign-up fields for a cooking club, styled only with OCS classes. The email field uses `type="email"` so the browser checks the format.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Cooking Club Signup Inputs | interactive: true -->
<!-- Each input is ocs__input plus one size modifier; the first also gets the gradient look. -->
<input type="text" class="ocs__input large gradient" placeholder="Favorite Dish to Cook">
<input type="text" class="ocs__input small" placeholder="First Name">
<input type="email" class="ocs__input medium" placeholder="Email Address">
```
