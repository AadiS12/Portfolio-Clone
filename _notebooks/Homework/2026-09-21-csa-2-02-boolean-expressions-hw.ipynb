## CSA 2.02 Boolean Expressions

A Boolean expression is a question Java answers with `true` or `false`, and the result can be stored in a `boolean` variable. `=` assigns a value, while `==` checks if two values are equal.

## Popcorn Hack: Fix the Two Bugs

**Bug 1:** `=>` isn't a Java operator, so it doesn't compile. It should be `>=`.  
**Bug 2:** `!=` asks "are they different?", but we want "exactly equal", so it should be `==`.

**Prediction:** `Bonus unlocked: true` and `Exact score: false`.

```java
// CODE_RUNNER: Popcorn hack - fix the two bugs so both lines print the expected values
public class BonusCheck {
    public static void main(String[] args) {
        int score = 120;
        int target = 100;

        // Bug 1 FIXED: "=>" is not a Java operator, so it did not compile. Greater-or-equal is ">=".
        boolean bonus = score >= target;

        // Bug 2 FIXED: "!=" asked "is it different?" The question is "is it EXACTLY equal?", which is "==".
        boolean exact = score == target;

        System.out.println("Bonus unlocked: " + bonus);
        System.out.println("Exact score: " + exact);
    }
}
BonusCheck.main(null);
```

## Homework Hack: Speed Check

Each `false` is replaced with a comparison. With `speed = 68`, `limit = 65`, `age = 17`:

**Predictions:** `speed > limit` → `true`, `speed == limit` → `false`, `age >= 18` → `false`, `speed % 2 != 0` → `false` (68 is even).

```java
// CODE_RUNNER: Homework - replace each false with a Boolean expression so the expected values print
public class SpeedCheck {
    public static void main(String[] args) {
        int speed = 68;
        int limit = 65;
        int age = 17;

        // 1. true when the driver is over the limit
        boolean speeding = speed > limit;        // 68 > 65

        // 2. true when the driver is going exactly the limit
        boolean exactLimit = speed == limit;     // 68 == 65 (== compares, = would assign)

        // 3. true when the driver is at least 18
        boolean adult = age >= 18;               // "at least" includes 18, so >= not >

        // 4. true when speed is an odd number (use %)
        boolean oddSpeed = speed % 2 != 0;       // odd numbers leave remainder 1 when divided by 2

        System.out.println("speeding: " + speeding);      // expected: true
        System.out.println("exactLimit: " + exactLimit);  // expected: false
        System.out.println("adult: " + adult);            // expected: false
        System.out.println("oddSpeed: " + oddSpeed);      // expected: false
    }
}
SpeedCheck.main(null);
```
