## SASS Buttons

All hacks use only OCS button classes and the `ocs__links` layout containers, with no custom CSS.

## Popcorn Hack 1: Add the Base Button Class

`ocs__btn` gives an element the shared OCS button look. An anchor is used because this action navigates to another page.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 — Base Button Class | interactive: true -->
<!-- ocs__btn supplies the standard button appearance. -->
<a class="ocs__btn" href="#">Browse the Study Guides</a>
```

## Popcorn Hack 2: Modifiers and Button Groups

The main action gets `accent pill fill`, the secondary action gets `outline`, and both sit inside an `ocs__links` group.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 — Modifiers and Button Group | interactive: true -->
<!-- ocs__links spaces and aligns related buttons. -->
<div class="ocs__links">
  <button class="ocs__btn accent pill fill">Start Practice Quiz</button>
  <a class="ocs__btn outline" href="#">Review Notes</a>
</div>
```

## Popcorn Hack 3: Status Tones and Wide Layout Bar

`alert-green` marks a positive action, `alert-red fill` marks the urgent one, and `ocs__links--wide` widens the bar.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 3 — Status Tones and Wide Layout | interactive: true -->
<!-- Tone classes carry meaning without hardcoded colors. -->
<div class="ocs__links ocs__links--wide">
  <button class="ocs__btn alert-green outline">Mark Lesson Complete</button>
  <a class="ocs__btn alert-red fill" href="#">Report a Problem</a>
</div>
```

## Homework Hack 1: Adaptive Resource Toolbar

The three categories use OCS utility classes only (no inline styles): red for urgent, green for self-care, accent outline for community.

```python
%%html
<!-- UI_RUNNER: Homework Hack 1 — Adaptive Resource Toolbar | interactive: true -->
<!-- Urgent items are filled red, self-care is filled green, community is a quieter outline. -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill" href="#">Urgent Support</a>
  <a class="ocs__btn alert-green fill pill" href="#">Self-Care</a>
  <a class="ocs__btn accent outline pill" href="#">Community</a>
</div>
```

## Homework Hack 2: Primary vs. Secondary Action Priority

One filled button is the clear primary action. The others use outlines so they stay visible without competing.

```python
%%html
<!-- UI_RUNNER: Homework Hack 2 — Primary vs Secondary Action Priority | interactive: true -->
<!-- fill = primary action; outline = secondary actions. -->
<div class="ocs__links">
  <button class="ocs__btn accent fill">Log Study Session</button>
  <a class="ocs__btn outline" href="#">View Weekly Progress</a>
  <a class="ocs__btn alert-red outline" href="#">Get Help Now</a>
</div>
```
