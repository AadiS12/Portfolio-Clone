## SASS Buttons

Completed all three Popcorn Hacks and both Homework Hacks using only OCS SASS button classes and layout containers.

## Popcorn Hack 1: Add the Base Button Class

Adding `ocs__btn` applies the shared OCS button styling with no custom CSS.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 — Base Button Class | interactive: true -->
<!-- ocs__btn supplies the standard OCS button look. -->
<!-- An anchor is the right element here because the button navigates to another page. -->
<a class="ocs__btn" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Browse Wellness Resources</a>
```

## Popcorn Hack 2: Modifiers and Button Groups

The first action gets the `pill` and `fill` modifiers, and both actions sit together in an `ocs__links` container.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 — Modifiers and Button Group | interactive: true -->
<!-- ocs__links lines up related actions with consistent spacing. -->
<!-- pill and fill make the primary action stand out from the secondary one. -->
<div class="ocs__links">
  <button class="ocs__btn accent pill fill">Start Today's Check-In</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Open Journal Notes</a>
</div>
```

## Popcorn Hack 3: Status Tones and Wide Layout Bar

The second action uses `alert-green`, the container becomes `ocs__links--wide`, and the crisis action uses `fill` as the highest-priority button.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 3 — Status Tones and Wide Layout | interactive: true -->
<!-- ocs__links--wide stretches the action bar into a wider responsive row. -->
<!-- Tone classes carry meaning (green = fine, red = urgent) with no hardcoded colors. -->
<div class="ocs__links--wide">
  <button class="ocs__btn alert-green outline">Mark Check-In Done</button>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Reach Crisis Support</a>
</div>
```

## Homework Hack 1: Adaptive Resource Toolbar

The three resource categories move from inline styling to the exact OCS utility classes the lesson requires.

```python
%%html
<!-- UI_RUNNER: Homework Hack 1 — Adaptive Resource Toolbar | interactive: true -->
<!-- Only OCS SASS utility classes appear below; there are no inline style attributes. -->
<!-- Red = urgent help, green = self-care, accent outline = the lower-priority community link. -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Immediate Help</a>
  <a class="ocs__btn alert-green fill pill" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Relax &amp; Recharge</a>
  <a class="ocs__btn accent outline pill" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Peer Groups</a>
</div>
```

## Homework Hack 2: Primary vs. Secondary Action Priority

Fill, outline, and alert modifiers build a clear visual hierarchy in the action row, again with no custom CSS.

```python
%%html
<!-- UI_RUNNER: Homework Hack 2 — Primary vs Secondary Action Priority | interactive: true -->
<!-- The filled accent button is the main action; outline buttons are visibly secondary. -->
<!-- alert-red with outline keeps the hotline easy to spot without outranking the main action. -->
<div class="ocs__links">
  <button class="ocs__btn accent fill">Record Today's Mood</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">See Monthly Patterns</a>
  <a class="ocs__btn alert-red outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Call the Helpline</a>
</div>
```
