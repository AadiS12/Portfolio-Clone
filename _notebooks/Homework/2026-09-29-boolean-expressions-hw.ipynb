## Python 3.05 Boolean Expressions

Three Popcorn Hacks, the MCQ, and an SFI backend validator.

## Popcorn Hack 1: Check One SFI Part Record

Each requirement is its own Boolean so it is clear which one passes.

```python
# CODE_RUNNER: Popcorn Hack 1 - Check SFI Part Fields
part = {
    "product_name": "Replacement Flywheels and Clutch Assemblies",
    "category": "Auto Racing",
    "spec_number": "1.1",
    "effective_date": "Nov. 9, 2001",
}
valid_categories = ["Auto Racing", "Drag Racing"]

print("Product name present:", len(part["product_name"].strip()) > 0)
print("Accepted category:", part["category"] in valid_categories)
print("Spec number present:", len(part["spec_number"].strip()) > 0)
print("Effective date present:", len(part["effective_date"].strip()) > 0)
```

## Popcorn Hack 2: Backend Create Rule

A part can be created only if every required field is valid **and** its spec number is **not** already stored. The first test is a duplicate and is rejected; the second uses a new spec number and is accepted.

```python
# CODE_RUNNER: Popcorn Hack 2 - Backend Create Rule
valid_categories = ["Auto Racing", "Drag Racing"]
stored_specs = ["1.1", "2.1"]

def can_create(part):
    fields_ok = all(part[key].strip() != "" for key in ["product_name", "spec_number", "effective_date"])
    category_ok = part["category"] in valid_categories
    is_new = part["spec_number"] not in stored_specs
    return fields_ok and category_ok and is_new

base = {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing",
        "spec_number": "1.1", "effective_date": "Nov. 9, 2001"}
fresh = dict(base, spec_number="1.3")

print("Duplicate spec accepted:", can_create(base))
print("New spec accepted:", can_create(fresh))
```

## Popcorn Hack 3: SFI Search Filter

A record matches if the query equals its name **or** its spec number, **and** it is in an accepted category.

```python
# CODE_RUNNER: Popcorn Hack 3 - SFI Search Filter
query = "1.2"
valid_categories = ["Auto Racing", "Drag Racing"]
records = [
    {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"},
]

matches = [
    r for r in records
    if (query == r["product_name"] or query == r["spec_number"]) and r["category"] in valid_categories
]

print("Matches found:", len(matches))
for r in matches:
    print(r["product_name"], "-", r["spec_number"])
```

## Homework Hack: SFI Backend Validator

Checks every required field, the category, and duplicates, then prints ACCEPT or REJECT for each test record.

```python
# CODE_RUNNER: 3.05 Homework - SFI Backend Validator
stored_specs = ["1.1", "2.1"]
valid_categories = ["Auto Racing", "Drag Racing"]

test_records = [
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.2", "effective_date": "Feb. 9, 2006"},
    {"product_name": "Replacement Flywheels", "category": "Street Car", "spec_number": "3.1", "effective_date": "Jan. 1, 2026"},
    {"product_name": "Existing Flywheel Record", "category": "Drag Racing", "spec_number": "1.1", "effective_date": "Jan. 1, 2026"},
]

def is_valid(record):
    has_name = record["product_name"].strip() != ""
    has_spec = record["spec_number"].strip() != ""
    has_date = record["effective_date"].strip() != ""
    good_category = record["category"] in valid_categories
    not_duplicate = not (record["spec_number"] in stored_specs)
    return has_name and has_spec and has_date and good_category and not_duplicate

for record in test_records:
    verdict = "ACCEPT" if is_valid(record) else "REJECT"
    print(verdict + ":", record["product_name"], "(" + record["spec_number"] + ")")
```

## Boolean Logic Explanation

`and` joins the rules because every one must pass. `not` flips the duplicate check so a record is accepted only when its spec number is not already stored. `or` is for alternatives, like matching either a product name or a spec number in a search.
