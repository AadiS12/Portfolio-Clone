## Popcorn Hack 1: Fix the Silent Failure

With `amount_requested = 0` the original printed nothing. That is a logic bug, not an error: the false case had no answer. Adding an `else` fixes it.

```python
# CODE_RUNNER: 3.6 Popcorn 1 - make both outcomes print
amount_requested = 0

# 1 condition -> 2 outcomes -> 2 answers
if amount_requested > 100:
    print("This message is asking for a large amount of money.")
else:
    print("This message is not asking for a large amount of money.")
```

## Popcorn Hack 2: Fix the `elif` Order

If `risk_score >= 3` comes first, every score of 3 or more matches it, so the `>= 6` and `>= 9` branches can never run. Checking the highest threshold first makes every branch reachable.

```python
# CODE_RUNNER: 3.6 Popcorn 2 - put thresholds in reachable order
risk_score = 9

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

## Homework Hack: Scam-Message Trainer

Each signal adds its own points. Then a highest-to-lowest `if / elif / else` chain gives every possible score an answer.

```python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it
def score_and_advise(message, amount_requested, known_contact):
    risk_score = 0

    # Separate ifs, no else: a missing signal just adds nothing.
    if "gift card" in message.lower():
        risk_score += 4
    if amount_requested > 100:
        risk_score += 3
    if not known_contact:
        risk_score += 2

    # 3 thresholds -> 4 outcomes -> 4 answers, highest first
    if risk_score >= 7:
        advice = "Do not reply. This message has multiple strong scam signals."
    elif risk_score >= 4:
        advice = "This message looks risky. Verify it before responding."
    elif risk_score >= 1:
        advice = "Use caution and check the sender or request."
    else:
        advice = "No scam signals were detected by this trainer."

    print("Risk score:", risk_score)
    print(advice)
    return risk_score

# Starter message from the lesson: all three signals are present.
score_and_advise("URGENT: send gift cards now to unlock your account", 500, False)
```

## Reachability Proof

Signals are worth 4 (gift card), 3 (amount over 100), and 2 (unknown sender), so each branch can be hit:

- **Score 9, highest branch:** gift card, amount `500`, unknown sender (`4 + 3 + 2`).
- **Score 4, second branch:** gift card only (`4`).
- **Score 2, third branch:** clean message from an unknown sender (`2`).
- **Score 0, else branch:** clean message from a known contact.

(Score 7 also lands in the top branch, for example gift card plus amount `500` from a known contact: `4 + 3`.)

```python
# CODE_RUNNER: 3.6 Homework tests - every reporting branch
def score_and_advise(message, amount_requested, known_contact):
    risk_score = 0
    if "gift card" in message.lower():
        risk_score += 4
    if amount_requested > 100:
        risk_score += 3
    if not known_contact:
        risk_score += 2
    if risk_score >= 7:
        advice = "Do not reply. This message has multiple strong scam signals."
    elif risk_score >= 4:
        advice = "This message looks risky. Verify it before responding."
    elif risk_score >= 1:
        advice = "Use caution and check the sender or request."
    else:
        advice = "No scam signals were detected by this trainer."
    print("Risk score:", risk_score)
    print(advice)

tests = [
    ("URGENT: send gift cards now", 500, False),  # 9
    ("Please buy a gift card", 0, True),          # 4
    ("Hi, checking in", 0, False),                # 2
    ("Hi Mom, see you Sunday", 0, True),          # 0
]
for message, amount, known in tests:
    print("---")
    score_and_advise(message, amount, known)
```
