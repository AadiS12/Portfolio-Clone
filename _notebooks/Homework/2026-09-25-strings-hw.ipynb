## 3.4 Strings: Interceptors

Indexing and slicing, string methods, concatenation, formatting, and decoding a CSV-style string.

## Popcorn Hack 1: Extract the Hidden Message

Slice the three hidden words out of the transmission.

```python
# CODE_RUNNER: Popcorn Hack 1 - Extract the hidden message using indexing and slicing
transmission = "xxSIGNALxxxCONFIRMEDxx77xxDELTA"

first = transmission[2:8]       # characters 2 up to (not including) 8
second = transmission[11:20]
third = transmission[-5:]       # last five characters

print("Word 1:", first)
print("Word 2:", second)
print("Word 3:", third)
print("Total length:", len(transmission))
print("Every other character:", transmission[::2])
```

## Popcorn Hack 2: Scrub and Search a Noisy Transmission

Trim, remove punctuation, lowercase, search, and count words.

```python
# CODE_RUNNER: Popcorn Hack 2 - Clean and search a noisy transmission
noisy = "   ...AGENT>>>Falcon---reports: the PACKAGE has ARRIVED!!!   "

step1 = noisy.strip()
print("Trimmed:", repr(step1))

step2 = step1
for junk in [">", "-", "!"]:
    step2 = step2.replace(junk, "")
print("Cleaned:", step2)

step3 = step2.lower()
print("Lowered:", step3)

print("Contains 'arrived':", "arrived" in step3)
print("Word count:", len(step2.split()))
```

## Popcorn Hack 3: Build a Decoded Report

Concatenation, `str()`, an f-string, and `join()`.

```python
# CODE_RUNNER: Popcorn Hack 3 - Build a decoded report
agent_name = "Falcon"
target_word = "MIDNIGHT"
confidence = 87
fragments = ["ALPHA", "SEVEN", "DELTA"]

headline = "Agent " + agent_name + " intercepted: " + target_word
print(headline)

print("Confidence: " + str(confidence) + "%")   # concatenation needs str()
print(f"Confidence: {confidence}%")             # f-string does it directly

joined = "-".join(fragments)
print("Joined fragments:", joined)

print(f"REPORT: {headline} | Confidence: {confidence}% | Fragments: {joined}")
```

## Homework Hack: Crack the Final Transmission

Split the CSV-style string, slice and replace parts of it, search it, and print Markdown and JSON-style output.

```python
# CODE_RUNNER: Homework - Crack the Final Transmission
raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

date, agent, message_code, confidence = raw_transmission.strip().split(",")
print("Fields:", date, agent, message_code, confidence)

print("Year:", date[:4])

message = message_code.replace("-", " ")
print("Decoded message:", message)

has_dawn = "dawn" in message.lower()
print("Mentions dawn:", has_dawn)

print(f"| {date} | {agent} | {message} | {confidence}% |")
print(f'{{"date": "{date}", "agent": "{agent}", "message": "{message}", "confidence": {confidence}, "contains_dawn": {str(has_dawn).lower()}}}')
```
