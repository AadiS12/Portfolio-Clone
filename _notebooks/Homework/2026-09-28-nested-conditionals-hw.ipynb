## Popcorn Hack 1: After-School Homework Plan

**Prediction for the starter values:** `Finish homework`. All three outcomes are tested below.

````python
# CODE_RUNNER: Popcorn Hack 1 - test every after-school homework outcome
for is_weekday, homework_finished in [(True, True), (True, False), (False, False)]:
    if is_weekday:
        if homework_finished:
            print("Free time")
        else:
            print("Finish homework")
    else:
        print("Weekend plan")
````

## Popcorn Hack 2: Repair the Student Account Check

The permission check wasn't indented inside the login check, so it ran even when logged out. Now it's nested, with an outer `else`.

````python
# CODE_RUNNER: Popcorn Hack 2 - repair and test the account check
for logged_in, has_permission in [(True, True), (True, False), (False, False)]:
    if logged_in:
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")
````

## Popcorn Hack 3: Lunch and Sports Practice

The lunch check comes first, and the practice check is nested inside it. All four combinations are tested.

````python
# CODE_RUNNER: Popcorn Hack 3 - test all lunch and practice paths
for lunch_packed in [True, False]:
    for practice_after_school in [True, False]:
        if lunch_packed:
            if practice_after_school:
                print("Bring packed lunch and an extra practice snack")
            else:
                print("Bring packed lunch")
        else:
            if practice_after_school:
                print("Buy lunch and get an extra snack for practice")
            else:
                print("Buy lunch")
````

## Homework Hack: Daily Routine Dashboard

Login is the outer check. Homework, lunch and practice are only checked after the student is logged in.

````python
# CODE_RUNNER: Homework Hack - high school daily routine dashboard
logged_in = True
homework_finished = False
lunch_packed = True
practice_today = True

if logged_in:
    if homework_finished:
        print("Homework complete - free time after school")
    else:
        print("Finish homework before free time")

    if lunch_packed:
        if practice_today:
            print("Bring packed lunch and an extra snack for practice")
        else:
            print("Bring packed lunch")
    else:
        if practice_today:
            print("Buy lunch and get an extra snack for practice")
        else:
            print("Buy lunch")
else:
    print("Log in first")
````

## Homework Tests

Tested logged out, unfinished homework with practice, finished homework without practice, and buying lunch.

````python
# CODE_RUNNER: Homework tests - exercise meaningful dashboard outcomes
tests = [
    (False, False, False, False),
    (True, False, True, True),
    (True, True, True, False),
    (True, True, False, True),
]

for logged_in, homework_finished, lunch_packed, practice_today in tests:
    print("Test:", logged_in, homework_finished, lunch_packed, practice_today)
    if logged_in:
        if homework_finished:
            print("  Homework complete")
        else:
            print("  Finish homework")
        if lunch_packed:
            if practice_today:
                print("  Packed lunch + practice snack")
            else:
                print("  Packed lunch")
        else:
            if practice_today:
                print("  Buy lunch + practice snack")
            else:
                print("  Buy lunch")
    else:
        print("  Log in first")
````

## Explanation

The other checks are nested inside the login check because a student shouldn't see their schedule until they're logged in. The practice check is nested inside the lunch check because the right lunch advice depends on both.
