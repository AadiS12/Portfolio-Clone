## Python 3.07 Nested Conditionals

Three Popcorn Hacks and a daily-routine dashboard.

## Popcorn Hack 1: After-School Homework Plan

**Prediction for the starter values:** `Finish homework`. The weekday branch is taken, then the inner `else` runs because the homework is not finished.

The three outcomes are tested below.

````python
# CODE_RUNNER: Popcorn Hack 1 - every after-school outcome
for is_weekday, homework_finished in [(True, True), (True, False), (False, False)]:
    if is_weekday:
        if homework_finished:
            print("Free time")
        else:
            print("Finish homework")
    else:
        print("Weekend plan")
````

## Popcorn Hack 2: Repair the Student Account Check

The starter was not nested: `if has_permission` lined up with `if logged_in`, so permission was checked even for logged-out students. Indenting it inside the login branch and adding the outer `else` fixes it.

````python
# CODE_RUNNER: Popcorn Hack 2 - repaired account check
for logged_in, has_permission in [(True, True), (True, False), (False, False)]:
    if logged_in:
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")
````

## Popcorn Hack 3: Lunch and Sports Practice

Lunch is checked first. Only inside that branch is practice checked. All four combinations are tested.

````python
# CODE_RUNNER: Popcorn Hack 3 - all lunch and practice paths
for lunch_packed in [True, False]:
    for practice_after_school in [True, False]:
        if lunch_packed:
            if practice_after_school:
                print("Bring packed lunch and an extra practice snack")
            else:
                print("Bring packed lunch")
        else:
            if practice_after_school:
                print("Buy lunch and get an extra snack for practice")
            else:
                print("Buy lunch")
````

The lunch condition is evaluated first, and the practice condition only runs inside the lunch branch that was chosen.

## Homework Hack: High School Daily Routine Dashboard

Login is the outer decision. Inside it, homework, lunch, and practice are checked, with nested decisions for meals and practice.

````python
# CODE_RUNNER: Homework Hack - high school daily routine dashboard
student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True,
}

if student["logged_in"]:
    print("Account access granted")

    if student["homework_finished"]:
        print("Homework complete: after-school time is available")
    else:
        print("Finish homework before free time")

    if student["lunch_packed"]:
        if student["practice_today"]:
            print("Bring packed lunch and an extra snack for practice")
        else:
            print("Bring packed lunch")
    else:
        if student["practice_today"]:
            print("Buy lunch and get an extra snack for practice")
        else:
            print("Buy lunch")

    if student["practice_today"]:
        if student["homework_finished"]:
            print("Go to practice after school")
        else:
            print("Plan homework time before practice")
    else:
        print("No practice today: use the extra time for schoolwork or rest")
else:
    print("Log in first before viewing the daily routine")
````

## Homework Tests

Five student states: logged out; logged in with unfinished homework and practice; finished homework and no practice (packed lunch); finished homework with practice (buying lunch); unfinished homework with no practice (buying lunch).

````python
# CODE_RUNNER: Homework tests - exercise the dashboard branches
def dashboard(student):
    if not student["logged_in"]:
        print("Log in first")
        return
    print("Homework complete" if student["homework_finished"] else "Finish homework")
    if student["lunch_packed"]:
        print("Packed lunch + practice snack" if student["practice_today"] else "Packed lunch")
    else:
        print("Buy lunch + practice snack" if student["practice_today"] else "Buy lunch")
    if student["practice_today"]:
        print("Go to practice" if student["homework_finished"] else "Plan homework before practice")
    else:
        print("No practice today")

states = [
    {"logged_in": False, "homework_finished": False, "lunch_packed": False, "practice_today": False},
    {"logged_in": True, "homework_finished": False, "lunch_packed": True, "practice_today": True},
    {"logged_in": True, "homework_finished": True, "lunch_packed": True, "practice_today": False},
    {"logged_in": True, "homework_finished": True, "lunch_packed": False, "practice_today": True},
    {"logged_in": True, "homework_finished": False, "lunch_packed": False, "practice_today": False},
]
for s in states:
    print("\nState:", s)
    dashboard(s)
````

## Explanation

Personal details should only appear after the account check passes, so login is the outer decision. The meal advice depends on the lunch choice first and then on practice, which is why those checks are nested.
