## Python 3.03 Math Expressions

This homework covers both Popcorn Hacks and a Bowling Alley Cost Report, all using variables and arithmetic operators instead of hardcoded answers.

## Popcorn Hack 1: Basic Arithmetic Operators

Use the provided variables with addition, subtraction, multiplication, and division instead of hardcoding the results.

```python
# CODE_RUNNER: Popcorn Hack - Basic Arithmetic Operators
value_a = 50
value_b = 5
value_c = 12

addition = value_a + value_b + value_c
subtraction = value_a - value_b
multiplication = value_b * value_c
division = value_a / value_b

print("Total:", addition)
print("Difference:", subtraction)
print("Product:", multiplication)
print("Division:", division)
```

**Expected results:** Total 67, Difference 45, Product 60, Division 10.0.

## Popcorn Hack 2: Movie Theater Booking Math

Multiplication, subtraction, and division give the seats taken, the seats remaining, and the seats left per screen.

```python
# CODE_RUNNER: Python Version of Popcorn Hack 2
total_seats = 150
seats_per_group = 6
groups_booked = 5

# Multiplication: seats used by all groups.
seats_taken = seats_per_group * groups_booked

# Subtraction: seats still open.
seats_left = total_seats - seats_taken

# Division: share the open seats evenly between two screens.
seats_per_screen = seats_left / 2

print("Total Seats Taken:", seats_taken)
print("Remaining Seats:", seats_left)
print("Open Seats per Screen:", seats_per_screen)
```

**Expected results:** Total Seats Taken 30, Remaining Seats 120, Open Seats per Screen 60.0.

## Homework Hack: Bowling Alley Cost Report

Calculate the weekly cost, the monthly cost, the monthly total with the one-time shoe-rental fee, and the average daily cost.

```python
# CODE_RUNNER: 3.03 HOMEWORK - Bowling Alley Cost Report
cost_per_game = 7
games_per_week = 3
weeks_per_month = 4
shoe_fee = 12

weekly_cost = cost_per_game * games_per_week
monthly_cost = weekly_cost * weeks_per_month
total_with_fee = monthly_cost + shoe_fee
average_daily_cost = monthly_cost / 30

print("Weekly Cost:", weekly_cost)
print("Monthly Cost:", monthly_cost)
print("Total Cost with Fee:", total_with_fee)
print("Average Daily Cost:", average_daily_cost)
```
