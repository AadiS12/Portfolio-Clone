## Python 3.03 Math Expressions

Both Popcorn Hacks and the Recovery Center Cost Report. Every result comes from the variables, so changing a starting value updates all the answers.

## Popcorn Hack 1: Basic Arithmetic Operators

Uses `+`, `-`, `*`, and `/` on the given variables. Note that `/` always returns a decimal in Python, so `100 / 10` is `10.0`.

```python
# CODE_RUNNER: Popcorn Hack - Basic Arithmetic Operators
value_a = 100
value_b = 10
value_c = 25

total = value_a + value_b + value_c      # add all three
difference = value_a - value_b           # A minus B
product = value_b * value_c              # B times C
quotient = value_a / value_b             # A divided by B

print("Total:", total)
print("Difference:", difference)
print("Product:", product)
print("Division:", quotient)
```

**Expected results:** Total `135`, Difference `90`, Product `250`, Division `10.0`.

## Popcorn Hack 2: Poway Recovery Center Booking Math

Multiplication gives the spots used, subtraction gives what is left, and division splits the remainder between two rooms.

```python
# CODE_RUNNER: Python Version of Popcorn Hack 2
total_monthly_capacity = 100
spots_requested_per_group = 4
sessions_booked = 3

# spots per session times number of sessions
total_spots_taken = spots_requested_per_group * sessions_booked

# what is left of the monthly capacity
remaining_capacity = total_monthly_capacity - total_spots_taken

# split evenly across two rooms
users_per_room = remaining_capacity / 2

print("Total Spots Taken:", total_spots_taken)
print("Remaining Capacity:", remaining_capacity)
print("Users per Room:", users_per_room)
```

**Expected results:** Total Spots Taken `12`, Remaining Capacity `88`, Users per Room `44.0`.

## Homework Hack: Recovery Center Cost Report

Weekly cost, monthly cost, monthly total with the one-time enrollment fee, and the average daily cost over a 30-day month.

```python
# CODE_RUNNER: 3.03 HOMEWORK - Recovery Center Cost Report
cost_per_session = 35
sessions_per_week = 3
weeks_per_month = 4
enrollment_fee = 75

weekly_cost = cost_per_session * sessions_per_week          # 35 * 3
monthly_cost = weekly_cost * weeks_per_month                # 105 * 4
total_cost_with_fee = monthly_cost + enrollment_fee         # fee is charged once
average_daily_cost = monthly_cost / 30                      # sessions only, not the fee

print("Weekly Cost:", weekly_cost)
print("Monthly Cost:", monthly_cost)
print("Total Cost with Fee:", total_cost_with_fee)
print("Average Daily Cost:", average_daily_cost)
```

## Homework Results

Weekly Cost: **105**  
Monthly Cost: **420**  
Total Cost with Fee: **495**  
Average Daily Cost: **14.0**
