## Popcorn Hack 1: Find the Level

**Prediction:** the first print shows the first value (`16` after my change), the second shows `20`, and the length is `3`. Only the first element changes, so the second value and the length stay the same.

```python
# CODE_RUNNER: 3.2 Popcorn 1 - find the level
# Lists keep order. Python starts counting positions at 0.
level_stars = [16, 20, 15]
print(level_stars[0])   # first element
print(level_stars[1])   # second element
print(len(level_stars)) # number of elements
```

## Popcorn Hack 2: Extend the Level Sequence

**Prediction:** `[6, 12, 12, 5]` and length `4`.  
**AP index:** AP indexes start at 1, so AP index `2` is the same element as Python index `1`. Changing an element keeps the length the same; `append()` adds one and the length goes up by one.

```python
# CODE_RUNNER: 3.2 Popcorn 2 - update and append
level_stars = [6, 9, 12]
level_stars[1] += 3        # update the second element
level_stars.append(5)      # add a new level at the end
print(level_stars)
print(len(level_stars))
```

## Homework Hack: Grow the Level Collection

Each element of `level_stars` is the stars from one level, in level order.

```python
# CODE_RUNNER: 3.2 Homework - update, append and traverse the level collection
level_stars = [12, 18, 10]

# Only update level two if it exists, so short lists cannot crash.
if len(level_stars) >= 2:
    level_stars[1] += 5

level_stars.append(7)

total_stars = 0
for stars in level_stars:     # works for any number of levels
    total_stars += stars

print(level_stars)
print(len(level_stars))
print(total_stars)
```

## Tests

Main list: `[12, 23, 10, 7]`, length `4`, total `52`. Empty list: length `0`, total `0`.

```python
# CODE_RUNNER: 3.2 Test - empty list
level_stars = []

if len(level_stars) >= 2:
    level_stars[1] += 5

total_stars = 0
for stars in level_stars:
    total_stars += stars

print(len(level_stars))
print(total_stars)
```

## Explanation

One list replaces a separate variable for every level, and the same loop adds up any number of levels, including none. Adding a level only needs `append()`.

## Design Thinking

**Need:** A game maker wants to add or reorder levels without managing many separate variables.  
**Goal:** Keep every level's stars in one ordered collection that can grow.  
**Representation:** Separate variables could hold the same numbers, but a list is easier to extend and loop over.  
**Prototype:** `level_stars` is updated at index 1, appended to, and traversed for the total.  
**Test/revision:** The main test gave four levels and a total of 52. The empty-list test gave 0 levels and a total of 0, which is why the update is guarded with a length check. Labels for each level would also help players who do not know list positions.
