## Python 3.15 Random Values

## Popcorn Hack 1: Random SFI Record Number

`random.randint(1, 4)` includes both ends, so `1`, `2`, `3`, and `4` can all appear. Repeated numbers are normal because each pick is independent.

```python
# CODE_RUNNER: Popcorn Hack 1 - Random SFI Record Number
import random

record_count = 4
for _ in range(5):
    print("Random record ID:", random.randint(1, record_count))
```

## Popcorn Hack 2: Random Structured SFI Car Part

Choosing the whole dictionary with `random.choice()` keeps each part's fields together.

```python
# CODE_RUNNER: Popcorn Hack 2 - Random SFI Part Record
import random

parts = [
    {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"},
]

part = random.choice(parts)
print("Product:", part["product_name"])
print("Category:", part["category"])
print("Spec number:", part["spec_number"])
```

## Popcorn Hack 3: Random SFI API Test

A random record and a random API action are chosen, then handled by the matching branch.

```python
# CODE_RUNNER: Popcorn Hack 3 - Random SFI API Test
import random

parts = [
    {"product_name": "Replacement Flywheels", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "spec_number": "2.1"},
]
stored_specs = ["1.1", "2.1"]

part = random.choice(parts)
action = random.choice(["GET/search", "POST/create", "PUT/update"])
print("Selected:", part["product_name"], "|", action)

if action == "GET/search":
    print("SEARCH RESULT:", part["product_name"], part["spec_number"])
elif action == "POST/create":
    if part["spec_number"] in stored_specs:
        print("CREATE REJECTED: duplicate spec", part["spec_number"])
    else:
        print("CREATE ACCEPTED:", part["spec_number"])
else:
    print("UPDATE TEST:", part["spec_number"])
```

## Homework Hack: SFI Backend QA Simulator

Five random tests across four backend actions, with duplicate-spec handling for creates.

```python
# CODE_RUNNER: 3.15 Homework - SFI Backend QA Simulator
import random

parts = [
    {"product_name": "Replacement Flywheels", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"},
]
actions = ["GET/search", "POST/create", "PUT/update", "DELETE/remove"]
stored_specs = ["1.1", "2.1"]

for test_number in range(1, 6):
    part = random.choice(parts)
    action = random.choice(actions)
    spec = part["spec_number"]
    print("\nTest", test_number)
    print("Record:", part["product_name"], "| Spec:", spec, "| Action:", action)

    if action == "GET/search":
        print("SEARCH: found", part["product_name"])
    elif action == "POST/create":
        if spec in stored_specs:
            print("CREATE REJECTED: duplicate spec number")
        else:
            stored_specs.append(spec)
            print("CREATE ACCEPTED: spec stored")
    elif action == "PUT/update":
        if spec in stored_specs:
            print("UPDATE ACCEPTED: record exists")
        else:
            print("UPDATE REJECTED: record does not exist")
    else:
        if spec in stored_specs:
            print("DELETE TEST: record would be removed")
        else:
            print("DELETE TEST: no matching record")
```

## Randomness Explanation

Record IDs stay inside the inclusive range, but repeats are expected because each pick is independent. Running a random program several times gives more evidence than one run, but no number of runs guarantees every record and action combination appears.
