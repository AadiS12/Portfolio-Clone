## Homework Hack: SFI Backend QA Simulator

Each run picks a random part and action, and `handle()` returns a backend-style result. Creating a spec adds it to the stored list, so creating it again is rejected as a duplicate.

```python
# CODE_RUNNER: Homework Hack - SFI Backend QA Simulator
import random

parts = [
    {
        "product_name": "Replacement Flywheels",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

actions = [
    "GET/search",
    "POST/create",
    "PUT/update",
    "DELETE/remove"
]

existing_spec_numbers = ["1.1", "2.1"]
test_count = 5

def handle(part, action, specs):
    spec = part["spec_number"]
    if action == "GET/search":
        if spec in specs:
            return "found " + part["product_name"]
        return "spec " + spec + " not found"
    elif action == "POST/create":
        if spec in specs:
            return "duplicate spec " + spec + " rejected"
        specs.append(spec)
        return "created spec " + spec
    elif action == "PUT/update":
        if spec in specs:
            return "updated spec " + spec
        return "cannot update missing spec " + spec
    else:
        if spec in specs:
            specs.remove(spec)
            return "removed spec " + spec
        return "cannot remove missing spec " + spec

specs = list(existing_spec_numbers)
for run in range(1, test_count + 1):
    part = random.choice(parts)
    action = random.choice(actions)
    print("Run", run, "| spec", part["spec_number"], "|", action, "->", handle(part, action, specs))

# Duplicate test: the same new spec created twice.
specs = list(existing_spec_numbers)
print(handle(parts[1], "POST/create", specs))
print(handle(parts[1], "POST/create", specs))
```

## Possible Results

There are 3 parts × 4 actions = **12** possible pairs, and a pair can repeat between runs. Creating spec 1.1 or 2.1 is always rejected as a duplicate. Spec 1.2 is created the first time and rejected the second time, as the duplicate test shows.
