## Homework Hack: Build the UESL Star Counter

A string for the game name, a Boolean for reduced motion, and three number variables for the level stars. The total starts at `40`. A bonus of `5` goes to level two, but the stored total stays `40` until it is calculated again, which gives `45`.

```python
# CODE_RUNNER: UESL 3.1 Homework - variables, assignment, and recalculation
game_name = "Star Trail"      # string
reduced_motion = True         # Boolean: on or off

level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

total_stars = level_one_stars + level_two_stars + level_three_stars
print("Game:", game_name)
print("Reduced motion:", reduced_motion)
print("Original total:", total_stars)

# The bonus changes level two, but total_stars still holds the old number.
bonus = 5
level_two_stars += bonus
print("Stored total after bonus:", total_stars)

# Assign again to recalculate the total from the new level values.
total_stars = level_one_stars + level_two_stars + level_three_stars
print("Recalculated total:", total_stars)
```

## Tests

**Bonus 5:** original `40`, stored `40`, recalculated `45`.  
**Bonus 0:** original `40`, stored `40`, recalculated `40`.

```python
# CODE_RUNNER: UESL 3.1 Tests - both required bonus cases
def run_case(bonus):
    one, two, three = 12, 18, 10          # fresh values for every case
    total = one + two + three
    original = total
    two += bonus
    stored = total                         # not updated yet
    total = one + two + three              # recalculated
    print("Bonus:", bonus, "| Original:", original, "| Stored:", stored, "| Recalculated:", total)

run_case(5)
run_case(0)
```

## Design Thinking

**Player need:** Some players find fast animation uncomfortable or hard to follow.  
**Goal:** Let the player turn on reduced motion while keeping the score data simple for the maker.  
**Representation:** A Boolean fits reduced motion because it is either on or off; integers fit the star counts.  
**Prototype:** Three level scores, a total, and a `reduced_motion` setting.  
**Test/revision:** Trying bonuses `5` and `0` showed that a stored total does not change by itself, so it must be recalculated whenever a level changes. Reduced motion stays a player choice because different players need different things.
